```python
import pandas as pd
# Load dataset
data = pd.read_csv("salary_data.csv")
# Inspect data
print(data.head())
print(data.columns)
print(data.shape)
# input features : age experience
# output features : income
```

```python
# Inputs (features)
X = data[["experience","age"]]
# Output (target)
y = data["income"]
```

```python
import numpy as np
# Number of features
n_features = X.shape[1]
# Initialize weights and bias
w = np.zeros(n_features)
b =0.0
# because weights define how nuch does each feature ccontribute in decision making
#to add a shift
# yess it could be risky as it may miss the features that contribute very little
```

```python
def predict(X, w, b):
    """
    X : input features
    w : weights
    b : bias
    returns predicted values
    """
    y_hat = X.dot(w) + b
    return y_hat

    # Because this model is performing linear regression, not classification.
    # it can take continous values
    # it is not used for classification
```

```python
def mean_squared_error(y, y_hat):
    """
    y     : actual values
    y_hat : predicted values
    """
    loss = ((y_hat - y) ** 2).mean()
    return loss
# To penalize larger errors more strongly
# Its error becomes much larger after squaring
# Absolute error is not differentiable at zero
```

```python
def compute_gradients(X, y, y_hat):
    """
    Compute gradients of MSE loss
    """
    N =len(y)
    # Gradient w.r.t weights
    dw = (2 / N) * X.T.dot(y_hat - y)
    db = (2 / N) * (y_hat - y).sum()

    return dw, db

    # because bias is constant
    # Because gradients are proportional to how wrong the prediction is
    # weights and bias stop updating
```

```python
def update_parameters(w, b, dw, db, lr):
    w = w - lr * dw
    b = b - lr * db
    return w, b
```

```python
lr = 0.0001
epochs = 1000

for epoch in range(epochs):
    y_hat = predict(X, w, b)

    loss = mean_squared_error(y, y_hat)

    dw, db = compute_gradients(X, y, y_hat)

    w, b = update_parameters(w, b, dw, db, lr)

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {loss}")

# yes, if the learning rate is appropriate, loss should gradually decrease
# learning rate is too high
# A smaller learning rate needs more epochs to converge, while a larger learning rate needs fewer epochs
```

```python
new_candidate = np.array([4.5,68])
predicted_salary = new_candidate.dot(w) + b
print(predicted_salary)
# yes because it change gradually
# Yes, the linear model produces smooth, continuous outputs without abrupt jumps.
# It avoids sudden discontinuities and provides more stable
```
